```python
print(torch.cuda.is_available(), torch.cuda.get_device_name(0) if torch.cuda.is_available() else "No GPU")
```

```python
model = YOLO("yolov8s.pt")

results = model.train(
    data=f"/content/find-ball/players-3/data.yaml",
    epochs=150,
    imgsz=1280,
    batch=64,
    patience=30,
    workers=8,
    cache="ram",
    amp=True,
    project=drive_path,
    name="detect_ball-players_augment_v1",
    exist_ok=True,
    plots=True,
    # --- AUMENTO DE DATOS ---
    hsv_v=0.6,           # Incrementa la variación de brillo (0.4 es por defecto) -> Clave para sombras
    hsv_s=0.6,           # Varía la saturación de color ante cambios de iluminación
    hsv_h=0.015,         # Ligera variación de tono
    mosaic=0.5,          # Junta 4 imágenes en una (rompe la memorización del fondo)
    mixup=0.1,           # Mezcla dos imágenes con opacidad para hacer los contornos más robustos
    degrees=10.0,        # Rotaciones ligeras para detectar la pelota en ángulos inclinados
    scale=0.5,           # Variación de escala para cambiar el tamaño visual del objeto
)
```
